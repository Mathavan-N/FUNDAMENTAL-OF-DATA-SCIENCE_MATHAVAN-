```python
 #Experiment 9 

import matplotlib.pyplot as plt
# Sample dataset
months = ["Jun", "Jul", "Aug", "Sep", "Oct"]
sales = [160, 190, 220, 170, 240]
expenses = [100, 130, 150, 110, 160]
# 1. Line Plot
plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Trend - Line Plot")
plt.grid(False)
plt.show()
# 2. Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(months, sales)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales - Bar Chart")
plt.grid(False)
plt.show()
# 3. Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(expenses, sales)
plt.xlabel("Expenses")
plt.ylabel("Sales")
plt.title("Sales vs Expenses - Scatter Plot")
plt.grid(False)
plt.show()
# 4. Histogram
plt.figure(figsize=(8, 5))
plt.hist(sales, bins=5, edgecolor="black")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Sales Distribution - Histogram")
plt.grid(False)
plt.show()
# 5. Pie Chart
plt.figure(figsize=(8, 5))
plt.pie(
    sales,
    labels=months,
    autopct="%1.1f%%"
)
plt.title("Sales Distribution - Pie Chart")
plt.show()
```
