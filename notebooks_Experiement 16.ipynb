```python
#Experiment 16 


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = {
    "total_bill": [12.50, 18.75, 25.40, 32.60, 15.80, 27.90, 21.50, 35.20, 19.40, 14.60],
    "tip": [1.50, 2.25, 3.80, 4.20, 2.00, 3.50, 2.80, 5.00, 2.50, 1.80],
    "sex": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "smoker": ["Yes", "No", "Yes", "No", "No", "Yes", "No", "Yes", "No", "Yes"],
    "day": ["Mon", "Mon", "Tue", "Tue", "Wed", "Wed", "Thu", "Thu", "Fri", "Fri"],
    "time": ["Lunch", "Dinner", "Lunch", "Dinner", "Lunch", "Dinner", "Lunch", "Dinner", "Lunch", "Dinner"],
    "size": [1, 2, 3, 4, 2, 5, 3, 6, 2, 4]
}
tips = pd.DataFrame(data)
print("Dataset:")
print(tips)
# Quantitative Analysis
print("\nQuantitative Analysis:")
print(tips.describe())
# Qualitative Analysis
print("\nQualitative Analysis:")
print(tips["time"].value_counts())
print(tips["day"].value_counts())
# Distribution of Total Bill
plt.hist(tips["total_bill"], bins=5, edgecolor="black")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.title("Distribution of Total Bill")
plt.show()
# Tip vs Total Bill
plt.scatter(tips["tip"], tips["total_bill"])
plt.xlabel("Tip")
plt.ylabel("Total Bill")
plt.title("Tip vs Total Bill")
plt.show()
# Tip vs Total Bill with trend line
plt.scatter(tips["tip"], tips["total_bill"])
m, b = np.polyfit(tips["tip"], tips["total_bill"], 1)
plt.plot(tips["tip"], m * tips["tip"] + b)
plt.xlabel("Tip")
plt.ylabel("Total Bill")
plt.title("Tip vs Total Bill with Regression Line")
plt.show()
# Box plot
plt.boxplot(tips["total_bill"])
plt.ylabel("Total Bill")
plt.title("Box Plot of Total Bill")
plt.show()
# Bar chart for Time
tips["time"].value_counts().plot(kind="bar")
plt.xlabel("Time")
plt.ylabel("Count")
plt.title("Customers by Time")
plt.show()
```
