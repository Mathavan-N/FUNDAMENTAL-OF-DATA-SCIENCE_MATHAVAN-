```python
# Experiment 15

import pandas as pd
import numpy as np
# New dataset
data = {
    "City": ["Coimbatore", "Hyderabad", "Pune", "Hyderabad", "Kolkata",
             "Coimbatore", "Pune", np.nan, "Kolkata", "Coimbatore"],
    "Age": [22, 30, 26, 36, 42, np.nan, 33, 48, 29, 39],
    "Income": [28000, 55000, 38000, np.nan, 72000,
               32000, 49000, 80000, 45000, 65000],
    "Selected": ["Yes", "No", "Yes", "No", "Yes",
                 "No", "Yes", "No", "No", "Yes"]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
# Handle missing values
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Income"] = df["Income"].fillna(round(df["Income"].mean()))
# Convert City into numerical columns
city = pd.get_dummies(df["City"])
# Create preprocessed dataset
updated_dataset = pd.concat(
    [city, df[["Age", "Income", "Selected"]]],
    axis=1
)
# Convert Selected into 0 and 1
updated_dataset["Selected"] = updated_dataset["Selected"].replace(
    {"No": 0, "Yes": 1}
)
print("\nPreprocessed Dataset:")
print(updated_dataset)
print("\nMissing Values:")
print(updated_dataset.isnull().sum())
```
