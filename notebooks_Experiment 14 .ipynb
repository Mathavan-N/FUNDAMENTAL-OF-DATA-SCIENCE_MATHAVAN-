```python
#Experiment 14 

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
# New dataset
data = {
    "Study_Hours": [1, 3, 5, 7, 9],
    "Test_Score": [40, 50, 70, 85, 100],
    "Attendance": [55, 65, 80, 90, 100]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
# Standard Scaling
sc = StandardScaler()
standard_scaled = sc.fit_transform(df)
print("\nStandard Scaled Data:")
print(standard_scaled)
# Min-Max Scaling
mms = MinMaxScaler()
minmax_scaled = mms.fit_transform(df)
print("\nMin-Max Scaled Data:")
print(minmax_scaled)
```
