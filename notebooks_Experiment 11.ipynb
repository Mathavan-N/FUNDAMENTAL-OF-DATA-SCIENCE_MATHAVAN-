```python
#Experiment 11


import pandas as pd
# Create employee dataset
data = {
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Arun", "Priya", "Kavin", "Divya", "Rahul"],
    "Salary": [45000, 58000, 52000, 75000, 68000]
}
df = pd.DataFrame(data)
# Display dataset
print("Employee Data:")
print(df)
# Display first rows
print("\nFirst Rows:")
print(df.head())
# Display information
print("\nData Information:")
df.info()
# Display salary column
print("\nSalary:")
print(df["Salary"])
# Basic statistics
print("\nSummary Statistics:")
print(df.describe())
# Salary statistics
print("\nAverage Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())
print("Maximum Salary:", df["Salary"].max())
print("Minimum Salary:", df["Salary"].min())
```
