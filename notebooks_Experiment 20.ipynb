```python
#Experiment 20 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
# New dataset
data = {
    "Annual_Income": [12, 14, 17, 19, 21, 30, 33, 36, 39, 43,
                      52, 56, 59, 63, 67, 76, 81, 86, 92, 98],
    "Spending_Score": [85, 78, 82, 88, 75, 65, 58, 62, 55, 60,
                       45, 38, 42, 35, 40, 25, 20, 28, 18, 15]
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
print("\nDataset Information:")
df.info()
print("\nFirst 5 Rows:")
print(df.head())
# Select features
features = df[["Annual_Income", "Spending_Score"]].values
# Create K-Means model
model = KMeans(n_clusters=4, random_state=42, n_init=10)
# Train the model
model.fit(features)
# Add cluster labels
df["Cluster"] = model.predict(features)
print("\nClustered Dataset:")
print(df)
# Plot clusters
plt.figure(figsize=(8, 6))
plt.scatter(
    df["Annual_Income"],
    df["Spending_Score"],
    c=df["Cluster"],
    s=100
)
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("K-Means Clustering")
plt.grid(False)
plt.show()
```
